```python
import numpy as np
#create a 4x4 matrix

z=np.array([[1,2,3,4],[3,2,1,5],[9,8,7,6],[4,5,3,2]])

print(z.ndim)

print(z.size)

print(z.shape)

print(z[0])

print(z[1,1])#indexing

print(z[1][1])#same as above

print(z[0:,1])#print second element from each row

print(z[0:,1:3])#print 2nd and 3rd element from all row

print(z[2,1:3])#3rd row 2nd and 3rd element




```

```python
#zero matrix

z=np.zeros([2,2],dtype=int)

z
```

```python
#one matrix
z=np.ones([3,3],dtype=int)
z
```

```python
#create gwo 2x2 matrix and store it in two variables

a=np.array([[1,2],[3,4]])
b=np.array([[5,6],[7,8]])
print(a)
print(b)
print(a+b)
print(a-b)
print(a*b)#normal multiplication
print(a@b)#proper matrix multiplication
print(a.dot(b))#also used for matrix multipllication
```

```python
#full matrix

z=np.full([3,3],4)
z
```

```python
#create a 4x4 matrix with value 9
z = np.full([4,4],9)
z
```

```python
#identity matrix
z=np.identity(3,dtype=int)
#OR
b=np.eye(4,dtype=int)
b
```

```python
#arange used like range
z=np.arange(1,10)
z
```

```python
#create a 4X3 matrix
z=np.ones([4,3],dtype=int).reshape(6,2)
z
```

```python
#create a 3X3 matrix with arange and reshape
z=np.arange(1,10).reshape(3,3)
z
```

```python
#create a 3x4 matrix with float values
z=np.array([[3.1,4.5,7.7,3.5],[2.9,6.3,5.5,8.5],[1.6,2.7,3.7,8.2]])
print(np.floor(z))
print(np.ceil(z))
print(np.round(z))#if .5 occurs the numbers is round off to nearest even number
print(z.max())
print(z.min())
print(z.sum())
```

```python
z=np.array([[3.1,4.5,7.7,3.5],[2.9,6.3,5.5,8.5],[1.6,2.7,3.7,8.2]])
print(z.max(axis=1))#axis=1 represent row in numpy
print(z.max(axis=0))#axis=0 represents column in numpy
print(z.sum(axis=1))
print(np.sqrt(z))
print(np.cos(z))
```

```python
z=np.array([[3.1,4.5,7.7,3.5],[2.9,6.3,5.5,8.5],[1.6,2.7,3.7,8.2]])
print(z)
print(np.sort(z))
print(np.sort(z,axis=0))
print(np.argsort(z,axis=1))#index based
```
